```python
from pathlib import Path
import json



# 1 Define the three haptic-feedback actions


HAPTIC_ACTIONS = {
    0: "Low",
    1: "Medium",
    2: "High"
}

print(
    "Haptic actions:"
)

print(
    HAPTIC_ACTIONS
)



# 2 Define the simulated parameters for each haptic action


HAPTIC_CONFIG = {
    0: {
        "label": "Low",
        "intensity": 0.25,
        "duration_ms": 150
    },

    1: {
        "label": "Medium",
        "intensity": 0.55,
        "duration_ms": 300
    },

    2: {
        "label": "High",
        "intensity": 0.85,
        "duration_ms": 500
    }
}

print(
    "\nHaptic configuration:"
)

for action, config in HAPTIC_CONFIG.items():

    print(
        action,
        config["label"],
        "Intensity:",
        config["intensity"],
        "Duration:",
        config["duration_ms"],
        "ms"
    )



# 3 Define a function for retrieving haptic settings


def get_haptic_config(
    action
):
    if action not in HAPTIC_CONFIG:
        raise ValueError(
            "Action must be 0, 1, or 2."
        )

    return HAPTIC_CONFIG[
        action
    ]



# 4 Check the haptic configuration using an example


example_action = 1

example_config = get_haptic_config(
    example_action
)

print(
    "\nExample action:",
    example_action
)

print(
    "Label:",
    example_config["label"]
)

print(
    "Intensity:",
    example_config["intensity"]
)

print(
    "Duration:",
    example_config["duration_ms"],
    "ms"
)



# 5 Check ordering of intensity and duration


intensities = [
    HAPTIC_CONFIG[action][
        "intensity"
    ]
    for action in [
        0,
        1,
        2
    ]
]

durations = [
    HAPTIC_CONFIG[action][
        "duration_ms"
    ]
    for action in [
        0,
        1,
        2
    ]
]

if not (
    intensities[0]
    < intensities[1]
    < intensities[2]
):
    raise ValueError(
        "Haptic intensity must increase "
        "from Low to High."
    )

if not (
    durations[0]
    < durations[1]
    < durations[2]
):
    raise ValueError(
        "Haptic duration must increase "
        "from Low to High."
    )

print(
    "\nHaptic ordering check passed."
)



# 6 Export the haptic-action configuration


BASE_DIR = Path.cwd()

OUTPUT_PATH = (
    BASE_DIR
    / "phase11_haptic_actions_rerun.json"
)

with open(
    OUTPUT_PATH,
    "w"
) as file:

    json.dump(
        HAPTIC_CONFIG,
        file,
        indent=4
    )

print(
    "\nSaved:",
    OUTPUT_PATH
)



# 7 Confirm Phase 11 completion


print(
    "\nPHASE 11 COMPLETE"
)

print(
    "Low:",
    HAPTIC_CONFIG[0]
)

print(
    "Medium:",
    HAPTIC_CONFIG[1]
)

print(
    "High:",
    HAPTIC_CONFIG[2]
)
```
